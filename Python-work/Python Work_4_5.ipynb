##  Factorisation de Cholesky

Si $A$ est une matrice symétrique et définie positive il est possible de construire une factorisation de la forme 

$\qquad A = L L^T$

où $L$ est une matrice triangulaire inférieure avec tous les coefficients de la diagonale positifs.

Dans le cas $n=2$ on a par identification 

$\qquad \left[
\begin{array}{rr}
a_{11} & a_{12} \\
a_{12} & a_{22} 
\end{array}\right]  = 
\left[
\begin{array}{rr}
\ell_{11} & 0 \\
\ell_{21} & \ell_{22}  
\end{array}\right]  \left[
\begin{array}{rr}
\ell_{11} & \ell_{21} \\
0 & \ell_{22}  
\end{array}\right]=
\left[
\begin{array}{rr}
\ell_{11}^2 & \ell_{11}\ell_{21} \\
\ell_{11}h_{21} & \ell_{21}^2 + \ell_{22}^2  
\end{array}\right]$

d'où  $\ell_{11} = \sqrt{a_{11}}$, $\ell_{21} = a_{12}/\ell_{11}$, $\ell_{22} = \sqrt{a_{22} - \ell_{21}^2}$. 

En généralisant au cas d'une matrice de taille $n$, on obtient l'algorithme suivant


$\quad$$h_{11} = \sqrt{a_{11}}$

$\quad$pour $i=2,...,n$

$\quad\quad$pour $j=1,...,i-1$

$\quad\quad\quad$ $\ell_{ij} = \displaystyle\frac{1}{\ell_{jj}}\left(a_{ij} - \displaystyle\sum_{k=1}^{j-1}\ell_{ik}\ell_{jk}\right)$

$\quad\quad$fin

$\quad\quad$$\ell_{ii} = \sqrt{a_{ii} - \displaystyle\sum_{k=1}^{i-1}\ell_{ik}^2}$


$\quad$fin

$\quad$retourner $L$

### Exercice 1. 

b) Programmer l'algorithme de la décomposition de Cholesky. On pourra compléter la fonction python ci-dessous :

```python
import numpy as np
def cholesky(A): 
    # A matrice carrée symétrique déf positive
    # la 
    dimA = A.shape
    n = dimA[0] 
    L = np.zeros([n,n])#... # initialization L comme matrice nulle
    L[0,0]=np.sqrt(A[0,0]) # initilization L[0,0]
    for i in range(1,n): # boucle  pour i  
        for j in range(0,i): # boucle  pour j :
            L[i,j]=(A[i,j] - np.dot(L[i,0:j],L[j,0:j]))/L[j,j]
        L[i,i]=np.sqrt(A[i,i]-np.dot(L[i,0:i],L[i,0:i]))
    return L

def cholesky2(A): 
    # A matrice carrée symétrique déf positive
    # la 
    dimA = A.shape
    n = dimA[0] 
    L = np.zeros([n,n])#... # initialization L comme matrice nulle
    L[0,0]=np.sqrt(A[0,0]) # initilization L[0,0]
    for i in range(1,n): # boucle  pour i  
        for j in range(0,i): # boucle  pour j :
            L[i,j]=(A[i,j] - np.dot(L[i,range(0,j)],L[j,range(0,j)]))/L[j,j]
        L[i,i]=np.sqrt(A[i,i]-np.dot(L[i,range(0,i)],L[i,range(0,i)]))
    return L
```

### Exercice 2 (extrait TP noté 22-23)

Soit A la matrice d'ordre 3 définie par 

$\quad A = \left[
\begin{array}{rrr}
1 & -1 & 1 \\
-1 & m & m \\
1 & m & m
\end{array}
\right]$

où $m=10^{-3}$. Soit $b=(1,-1,1)$. On considère le systèle $Ax = b$

1) Ecrire, ci-dessous, les tableaux (arrays) numpy avec la matrice $A$ et le vecteur $b$. 
On pourra compléter les instructions ci-dessous :

```python
# compléter les espace marqués avec " ... "
m = 1.e-3;
A = np.array([1, -1, 1, -1, m, m, 1, m, m],dtype=float).reshape(3,3)
b = np.array([1,-1,1],dtype = float)
print("A = ", A)
print("b = ", b)
```

2) Avec solve de numpy, résoudre  ci-dessous le système linéaire $Ax = b$. Garder la solution dans un tableau $x$

```python
# compléter les espace marqués avec " ... "

#calcul solution
x = np.linalg.solve(A,b)
# affichage solution
print("solution solve x = ")
print(x)
```

4) Avec solve de numpy, résoudre  ci-dessous le système linéaire $Ax = c$. Garder la solution dans un tableau $z$
