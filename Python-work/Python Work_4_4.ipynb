##   Méthode de Gauss avec pivot

Dans cette partie on va présenter une technique qui permet de calculer la factorisation LU de toute matrice non-singulière même dans le cas où la matrice ne vérifie pas les conditions de la proposition vue en cours (pour rappel, la factorisation LU d'une matrice $A$ existe si toutes les matrices principales $A_k$ de $A$ sont non-singulières).

Cette technique (dite du pivot partiel) est basée sur la permutation des lignes de $A$ afin d'avoir le pivot de plus grande taille possible (en valeur absolue). 

On considère le système $Ax = b$ avec $A$ une matrice d'ordre $n$. 
Dans une première version, on calculera la seulement la matrice $U$ et le vecteur $\tilde{b}$ tels que 
$Ax= b$ est équivalent à $Ux = \tilde{b}$ (système échelonné)

Une version de l'algorithme (sans resolution du système) est la suivante : 


$\quad$pour $k=1,...,n-1$

$\quad\quad$trouver $r\in\{k,\dots,n\}$ tel que $|a_{rk}| = \max\{|a_{ik}|,i\in\{k,\dots,n\}\}$.

$\quad\quad$permuter les lignes $k$ et $r$ de $A$ et de $b$.

$\quad\quad$pour $i=k+1,...,n$

$\quad\quad\quad$ $m = a_{ik}/a_{kk}$ 

$\quad\quad\quad$ $a_{ik} = 0 $

$\quad\quad\quad$pour $j=k+1,...,n$

$\quad\quad\quad\quad$$a_{ij} = a_{ij} - m \times a_{kj}$

$\quad\quad\quad$fin (boucle $j$)

$\quad\quad\quad$$b_i = b_i - m \times b_k$ 

$\quad\quad$fin (boucle $i$)

$\quad$fin (boucle $k$)


Obs. A la fin de la boucle pour $k$, la matrice $A$ correspondra à $U$ et le vecteur  $b$ à $\tilde{b}$. Une fois l'échelonnement éfectué, on pourra  calculer la solution du système $Ax = b$ en utilisant la méthode de remontée (pour la résolution de systèmes triangulaires supérieurs). 

Remarques. 

1) Pour calculer l'indice $r$ de la ligne contenant le pivot on pourrait utiliser la fonction numpy.argmax, 

2) Pour permuter les lignes $k$ et $r$ de la matrice on pourrait utiliser les instructions suivantes : 

   ind = np.arange(0,n)  # les indices de 0 à n-1
   
   ind[k] = r            # on permute les indices k et r dans ind
   
   ind[r] = k
   
   A = A[ind,:]          # on permute les lignes de A



```python
# Exemple 1 
import numpy as np
v=np.array([1.0,-2.0, -2.1, 1.1, 2.09])
r = np.argmax(np.abs(v))
print("r = ", r, ", v[r] = ", v[r], ", |v[r]| = ", np.abs(v[r]))
```

a) Programmer une fonction python qui effectue la permutation des lignes $k$ et $r$ d'une matrice $A$ ou un vecteur. Compléter la fonction ci-dessous

```python
def permutation_lignes(A,k,r):
    # on effectue la permutation de lignes k et r de A
    # A matrice de taille n lignes et m colonnes ou vecteur de taille n
    # k,r indices entre 0 et n-1
    # on retourne la matrice A avec la permutation
    dimA = A.shape
    sizeA = A.size
    n = dimA[0]  # nombre de lignes     
    ind = np.arange(0,n)  # les indices de 0 à n-1
    ind[k] = r             # on permute les indices k et r dans ind
    ind[r] = k 
    if n != sizeA: # A est une matrice
        A = A[ ind , : ]
    else :           # A est un vecteur
        A = A[ind]
    return A
```
