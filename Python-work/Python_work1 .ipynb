Listes et arrays (tableaux)

- on peut créer une liste vide avec nomliste=[] 
- avec x.append() on peut ajouter un nouveau élément à la liste. 
- le i-ème élément de la liste est nomliste[i] (attention, on commence la numérotation à 0)
- len(nomliste) donne la taille de la liste
- avec nomliste.reverse() on inverse l'ordre de la liste 



```python
#Exemple
liste=[]
for i in range(1,11):
    liste.append(1/i)
print(liste)

#le premier élément de la liste : 
print("premier élément = ", liste[0])
#le dernier élément de la liste : 
print("dernier élément = ", liste[9])
# la taille de la liste
print("taille = ", len(liste))
```

Arrays (tableaux)


- avec nomarray = numpy.zeros(N), on  crée un array nomarray rempli de 𝑁 zeros (donc 𝑁 c'est la taille du tableau) 

- si $0 \le k \le N-1$, alors avec nomarray[k] = V on affecte le  
k-ième élément du tableau avec la valeur V

- nomarray.size donne la taille du tableau.



```python
import numpy #on charge numpy (si ce n'était pas fait avant)
N=10
tableau = numpy.zeros(N)
print(tableau)
```

```python
for i in range(0,10):
    tableau[i] = i**2
print(tableau)
```

```python
# si on dépasse la taille du tableau, il y a une erreur : 
for i in range(0,11):
    tableau[i] = i**2
print(tableau)
```

```python
tableau2 = numpy.zeros(N)
for i in range(0,10):
    tableau2[i] = (9-i)**2
print(tableau2)
```

```python
# les opérations classiques entre tableaux  sont définies 
somme = tableau + tableau2
print(somme)
produit = tableau * tableau2 # produit des entrées du tableau
print(produit)
racine = tableau**0.5
print(racine)
```

## Exercice 1

Soit $N$ un nombre entier naturel. L'algorithme suivant calcule la représentation de $N$ en base 2 : 

Initialisation :  $q = N$, Nb2=[]  

tant que $q \neq 0$ faire 

      r  = reste de la division entière de q par 2
      
      q  = quotient de la division entière de q par 2

      Ajouter r dans la liste Nb2

fin tant que

Inverser la liste Nb2

Afficher la liste

On pourra compléter le programme suivant : 

```python
N=input("nombre entier = ")
N=int(N)
q=N;
Nb2 = []
while q != 0:
    r = q%2
    Nb2.append(r) 
    q=q//2
print(Nb2)
Nb2.reverse()
print('N = ',Nb2,'_2')
```

## Exercice 3

En utilisant les programmes des exercice 1 et 2, écrire une fonction python ayant comme argument un nombre $x$ (réel) et qui devra retourner  un tableau composé de $s$, le signe de $x$ ($s=0$ si $x$ est positif ou nul, $s=1$ si $x$ est negatif), du tableau avec la représentation en base $2$ de la partie entière de $|x|$ et du tableau avec la représentation de la partie décimale de $|x|$.

Ind. Pour trouver la partie entière de $x$ on peut utiliser la fonction numpy "floor" : numpy.floor() ou (np.floor() si on a fait "import numpy as np" avant). Attention la réponse est de type float et pas int.

On pourra compléter la fonction suivante : 

```python
import numpy as np

def repbin(x):
    if x >= 0: 
        s=0
    else:
        s=1
    N = np.floor(abs(x))
    f = abs(x)-N
    N=int(N)
    q=N;
    Nb2 = []
    while q != 0:
        r = q%2
        Nb2.append(r) 
        q=q//2
    #print(Nb2)
    Nb2.reverse()
    fn=f
    count = 0
    fb2 = []
    while fn != 0 and count < 20:
        fn=2*fn 
        if fn >= 1 : 
            d = 1
            fn = fn - 1
        else : 
            d = 0
    #    print('d_',count+1,' = ',d)
        fb2.append(d)
        count += 1
    #print(fb2)
    return [s,Nb2,fb2]
```
