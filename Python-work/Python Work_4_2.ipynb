a) Construire la matrice de Hilbert pour $n=4$ (voir TP4 partie 1, Exercice 3). Avec np.ones(), construire le vecteur $x= (1,1,1,1)$ puis calculer $b= Hx$. Résoudre, avec np.linalg.solve, le système $Hx = b$. Comparer avec la solution exacte x = (1,1,1,1).

b) Refaire a) avec $n=7$, $n=10$, $n=15$. Pour chaque cas, calculer  l'erreur entre la solution exacte $x$ et la solution calculée par numpy $x_{np}$

$ || x - x_{np} || $ 

où $|| \cdot ||$ corréspond à la norme euclidienne. 



```python
import numpy as np
n=4
def mathilbert(n):
    H = np.zeros((n,n)) # initialisation avec un matrice nule de taille nxn
    for i in range(0,n): # compléter la paranthèse  
        for j in range(0,n):#... à compléter
            H[i,j] = 1/(i+j+1) # car i,j >= 0, on decale de 1, et (i+1)+(i+j)-1 = i+j+1
    return H

H = mathilbert(n) #fonction mathilbert Exercice 3 , TP4 partie 1
x = np.ones(n) # à compléter
# suite de l'exercice...
b = H@x 
print("second membre b = ", b)
xnp = np.linalg.solve(H,b)
print("solution solve numpy de Hx = b : xnp = ", xnp)
```

b) Refaire a) avec $n=7$, $n=10$, $n=15$. Pour chaque cas, calculer  l'erreur entre la solution exacte $x$ et la solution calculée par numpy $x_{np}$

$ || x - x_{np} || $ 

où $|| \cdot ||$ corréspond à la norme euclidienne. 


```python
for n in [4,7,10,15]:
    H = mathilbert(n) 
    x = np.ones(n) 
    b = H@x 
    print("b = ", b)
    xnp = np.linalg.solve(H,b) 
    print("n = ",n," solution solve numpy : xnp = ", xnp)
    erreur=np.sqrt(np.dot(x-xnp,x-xnp))
    print("n = ",n," erreur ||x-xnp|| = ", erreur)
    erreur=np.linalg.norm(x-xnp)
    print("n = ",n," erreur2 ||x-xnp|| = ", erreur)    
```

## Exercice 2

On veut construire la matrice $n \times n$ triangulaire supérieure suivante 

$\quad  U = \left(\begin{array}{rrrrrr}
              n & n-1 & n-2 & \dots & \dots & 1 \\
              0 &  n & n-1 & n-2 & \dots & 2  \\
              0 & 0 & n  & n-1 &\ddots & 3 \\
              \vdots & \ddots & \ddots & \ddots & \ddots & \vdots \\ 
              0 & \dots & \dots & 0 & n & n-1 \\
              0 & \dots & \dots &  \dots & 0 & n  
            \end{array}
\right).
$

Ecrire une fonction python pour construire des matrices de ce type. Utiliser le moins de boucles for possible. 
Tester avec $n=5,10,15$. 

Indication : on pourrait utiliser la fonction np.diag()

```python
def triangulairesup(n):
    U=np.zeros((n,n))
    for i in range(0,n):
        U = U + np.diag((n-i)*np.ones(n-i),i)
    return U

#test avec n=4
n=4
U = triangulairesup(n)
print("n = ", n)
print(U)

# test avec n=10
n=10
U = triangulairesup(n)
print("n = ", n)
print(U)
```

```python
#une autre manière : avec np.arange
def triangulairesup(n):
    U=np.zeros((n,n))
    for i in range(0,n):
        U[i,i:n] = np.arange(n,i,-1) #on remplie les coefficients de la ligne i à partir de la diagonale
    return U

# test de np.arange avec l'argument "-1" : 
print(np.arange(10,2,-1))
# test triangulairesup 
U = triangulairesup(4)
print(U)
```

### Résolution de systèmes triangulaires inférieurs (méthode de déscente) 

Soit  $L$ une matrice $n\times n$ triangulaire 
inférieure non-singulière (c.-à-d., $\ell_{ij} = 0$ si $j>i$ et $det(L)\neq 0$). 
Soit $b$ un vecteur de taille $n$. 
Pour résoudre le système $Lx = b$ 
on peut utiliser l'algorithme 
suivant  : 

1. Etant donnés une matrice $L$ (triangulaire inférieure) de taille $n\times n$ et un vecteur $b$ de taille $n$

2. $x_1 = \displaystyle\frac{b_1}{\ell_{11}}$ 

3. pour $i = 2,\dots,n$

$\quad\quad\quad x_i=\displaystyle\frac{1}{\ell_{ii}}\left(b_i 
	  - \displaystyle\sum_{j=1}^{i-1} \ell_{ij}x_j \right)$

$\quad\quad\!\!$fin
  
4. Retourner $x$

### Résolution de systèmes triangulaires supérieurs (méthode de remontée) 

Soit $U$ 
une matrice $n\times n$ triangulaire 
supérieure non singulière (c.-à-d., $u_{ij} = 0$ si $j<i$ et $det(U)\neq 0$). Soit $b$ un vecteur de taille $n$.  
Pour résoudre le système $Ux = b$ 
on peut utiliser l'algorithme 
suivant : 

1. Etant donnés une matrice $U$ (triangulaire supérieure) de taille $n\times n$ et un vecteur $b$ de taille $n$

2. $x_n = \displaystyle\frac{b_n}{u_{nn}}$

3. pour $i = n-1,\dots,1$

$\quad\quad\quad x_i = \displaystyle\frac{1}{u_{ii}}\left(b_i 
	  - \displaystyle\sum_{j=i+1}^n u_{ij}x_j \right)$

$\quad\quad$fin

4. retourner $x$
