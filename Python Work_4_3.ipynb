##  Méthode de Gauss 

On veut résoudre $Ax = b$ ($A$ matrice 
 $n\times n$ non-singulière, $b,x\in \mathbb{R}^n$). Avec des transformations 
élémentaires et sous certaines conditions sur $A$ on peut transformer 
le système original dans un système équivalent $Ux = \tilde{b}$ avec $U$ une 
matrice triangulaire supérieure.  

Une version de l'algorithme (sans resolution du système) est la suivante : 


$\quad$pour $k=1,...,n-1$

$\quad\quad$ si $|a_{kk}| < \epsilon$ alors  

$\quad\quad\quad$ message d'erreur (pivot nul), sortir

$\quad\quad$pour $i=k+1,...,n$

$\quad\quad\quad$ $m = a_{ik}/a_{kk}$ 

$\quad\quad\quad$ $a_{ik} = 0 $

$\quad\quad\quad$pour $j=k+1,...,n$

$\quad\quad\quad\quad$$a_{ij} = a_{ij} - m \times a_{kj}$

$\quad\quad\quad$fin (boucle $j$)

$\quad\quad\quad$$b_i = b_i - m \times b_k$ 

$\quad\quad$fin (boucle $i$)

$\quad$fin (boucle $k$)


Observations : $\epsilon$ est une valeur petite à fixer (c'est-à-dire, une tolérance, par exemple $10^{-14}$). 
Dans cette version, la matrice $A$ et le vecteur $b$ sont modifiés à chaque étape. 

A la fin de la boucle pour $k$, la matrice $A$ correspondra à $U$ et le vecteur  $b$ à $\tilde{b}$. Une fois l'échelonnement éfectué, on pourra  calculer la solution du système $Ax = b$ en utilisant la méthode de remontée (pour la résolution de systèmes trinagulaires supérieurs).

```python
import numpy as np


import numpy as np

def gauss(A,b):
    # algorithme de Gauss pour échelonnement de (A|b)
    # avec A matrice nxn et b vecteur de taille n
    tol = 1e-14
    dimA = A.shape
    n=dimA[0]
    m=dimA[1]
    if n!=m: #on vérifie si la matrice est carrée 
        print("matrice non carrèe")
        return
    for k in range(0,n-1): 
        if abs(A[k,k])< tol:
            print("pivot (presque) nul, arrêt", A[k,k])
            return
        for i in range(k+1,n):
            m = A[i,k]/A[k,k]
            A[i,k]=0
            for j in range(k+1,n):
                A[i,j] = A[i,j] - m*A[k,j]
            #fin boucle j
            b[i]=b[i] - m*b[k]
        #fin boucle i
    #fin boucle k        
    return [A,b]
#fin fonction gauss
```

### Méthode de décomposition LU 

A partir de 
la méthode de Gauss et sous certaines conditions on peut factoriser $A$ comme un produit entre deux matrices : $A=LU$, avec $U$ la matrice triangulaire supérieure obtenue avec la méthode de Gauss, et $L$ une matrice triangulaire inférieure qui contient les multiplicateurs $m_{ij}$.

Dans la  version ci-dessous, la matrice $A$ n'est pas modifiée et les matrices $L$ et $U$ sont calculées séparément : 

$\quad$Initialisation $L=I$, $U=A$ 

$\quad$pour $k=1,...,n-1$  

$\quad$$\quad$ si $|u_{kk}| < \epsilon$ alors  

$\quad$$\quad$$\quad$ message d'erreur (pivot nul), sortir


$\quad$$\quad$pour $i=k+1,...,n$    

$\quad$$\quad$$\quad$$m = u_{ik}/u_{kk}$ 

$\quad$$\quad$$\quad$$\ell_{ik} = m $  

$\quad$$\quad$$\quad$$u_{ik} = 0 $  

$\quad$$\quad$$\quad$pour $j=k+1,...,n$   

$\quad$$\quad$$\quad$$\quad$$u_{ij} = u_{ij} - m \times u_{kj}$ 
                                

$\quad$$\quad$$\quad$fin              

                     
$\quad$$\quad$fin              

$\quad$fin   

$\quad$retourner $L$ et $U$

### Exercice 2. 

a) Programmer ci dessous une fonction python de l'algorithme de factorisation LU ci-dessus. Pour initialiser les matrices $L$ et $U$ on pourra utiliser np.eye (pour créer la matrice identité) et np.copy (pour copier la matrice A), respectivement.


```python
def lu(A):
    tol = 1e-14
    dimA = A.shape
    n=dimA[0]
    m=dimA[1]
    if n!=m:
        print("matrice non carrèe")
        return
    L = np.eye(n) #initialisation L=I (identité)
    U = np.copy(A)#initialisation U=A (copie de A)
    for k in range(0,n-1): 
        if abs(U[k,k])<tol:
            print("pivot (presque) nul, arrêt")
            return
        for i in range(k+1,n):
            m = U[i,k]/U[k,k]
            U[i,k]=0
            L[i,k]=m
            for j in range(k+1,n):
                U[i,j] = U[i,j] - m*U[k,j]
            #fin boucle j
        #fin boucle i
    #fin boucle k        
    return [L,U]
#fin fonction lu

```

### Deuxième version pour la  décomposition LU 


Avec l'algorithme ci-dessous on calcule la décomposition LU en modifiant la matrice A pour garder dans sa partie 
triangulaire inférieure les multiplicateurs $m$ (donc la matrice L) et dans sa partie supérieure la matrice U (Noter que la diagonale de $L$ n'est pas gardé) : 


$\quad$pour $k=1,...,n-1$  

$\quad$$\quad$ si $|a_{kk}| < \epsilon$ alors  

$\quad$$\quad$$\quad$ message d'erreur (pivot nul), sortir


$\quad$$\quad$pour $i=k+1,...,n$    

$\quad$$\quad$$\quad$$m = a_{ik}/a_{kk}$ 

$\quad$$\quad$$\quad$$a_{ik} = m $  

$\quad$$\quad$$\quad$pour $j=k+1,...,n$   

$\quad$$\quad$$\quad$$\quad$$a_{ij} = a_{ij} - m \times a_{kj}$ 
                                

$\quad$$\quad$$\quad$fin              

                     
$\quad$$\quad$fin              

$\quad$fin   

Observations : On peut voir que la matrice $A$ est modifiée à chaque étape.  
A la fin de la boucle pour $k$, la partie triangulaire supérieure (avec la diagonale) 
correspondra à $U$. Aussi, pour éviter l'utilisation d'une autre matrice, 
les multiplicateurs  $m$ sont stockés à la place des entrées 
$a_{ij}$, de la matrice $A$, qui deviennent $0$ à chaque étape. Ainsi, la matrice $L$ (qui a 
une diagonale principale égale à $1$)  correspondra à la partie 
triangulaire inférieure de $A$ (sans la diagonale).

```python

def lu2(A):
    # A matrice de taille nxn
    tol = 1e-14
    dimA = A.shape
    n=dimA[0]
    m=dimA[1]
    if n!=m:
        print("matrice non carrèe")
        return
    for k in range(0,n-1): 
        if abs(A[k,k])<tol :
            print("pivot (presque) nul, arrêt")
            return
        for i in range(k+1,n):
            m = A[i,k]/A[k,k]
            A[i,k] = m 
            for j in range(k+1,n):
                A[i,j] = A[i,j] - m*A[k,j]
            #fin boucle j
        #fin boucle i
    #fin boule k            
    return A
```

c) Ecrire une fonction python pour résoudre un système $Ax=b$  en utilisant la fonction lu2, et les fonctions remontée et descente du TP précédent. (Si $A=LU$ alors $Ax=b$ est équivalent à $L(Ux)= b$. Si on pose $Ux=y$, alors pour calculer $x$ on peut résoudre d'abord le système $Ly = b$ puis le système $Ux = y$).




```python
# corrigé fonction descente 
def descente(L,b):
    dimL=L.shape
    dimb=b.size
    n=dimL[0]
    m=dimL[1]
    if(n != m):
        print("matrice non carrée !")
        return
    if(dimb != n):
        print("dimensions de L et b non compatibles ")
        return
    if(np.prod(np.diag(L)) == 0):
        print("matrices L singulière ")
        return
    x=np.zeros(n)
    x[0]=b[0]/L[0,0]
    for i in range(1,n):
        x[i] = (b[i] - np.dot(L[i,range(0,i)],x[0:i]))/L[i,i]
    return x

#fin descente
  
def resolutionLU(A,b):
    # A matrice nxn
    # b matrice colonne nx1 (ou ligne)
    # a compléter ...
    dimA=A.shape
    n=dimA[0]
    m=dimA[1]  
    lu2(A)
    # on recupère les matrices L et U de A :
    L=np.zeros((n,n))
    U=np.zeros((n,n))
    for i in range(0,n):
        U += np.diag(np.diag(A,i),i) 
        if i==0:
            L = np.diag(np.ones(n))
        else :
            L += np.diag(np.diag(A,-i),-i)
    # on résout les systèmes Ly = b et Ux = y
    y=descente(L,b)
    x=remontee(U,y)
    return x


```

Ecrire une fonction python qui effectue la permutation des lignes $k$ et $r$ d'une matrice ou un vecteur. 
Compléter la fonction ci-dessous

```python

def permutation_lignes(A,k,r):
    #version 1
    # on effectue la permutation de lignes k et r de A
    # A matrice de taille n lignes et m colonnes ou vecteur de taille n
    # k,r indices entre 0 et n-1
    # on retourne la matrice (ou vecteur) A modifiée avec la permutation
    # on assume que les indices k et r sont entre 1 et n 
    kk = k-1
    rr = r-1 
    ligneAk = np.copy(A[kk,:])
    A[kk,:] = A[rr,:]
    A[rr,:] = ligneAk
    return A
```
