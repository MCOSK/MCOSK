```python
import numpy as np
```

Création d'une matrice et d'un vecteur (enlever les commentaires des lignes d'affichage et exécuter) :

Pour connaître les dimensions d'une matrice on peut utiliser shape, par exemple (enlever les commentaires des lignes d'affichage et exécuter) : 

```python
print(A.shape)
```

```python
L = np.arange(0,8).reshape(4,2)
print(L)
print(L.shape) 
print("nombre de lignes n = ", L.shape[0])
print("nombre de lignes m = ", L.shape[1])
```

Pour avoir accès au coefficient $a_{i,j}$ d'une matrice on utilise la syntaxe A[i,j]. Attention : si la matrice est de taille $n \times m$, les indices  vont de $0$ à $n-1$ pour i et de $0$ à $m-1$ pour $j$  (enlever les commentaires des lignes d'affichage et exécuter) :

```python
print(L[0,0])
print(L[3,1])

n = L.shape[0]
m = L.shape[1]
# on peut parcourir tous les coefficients avec deux boucles for : 
for i in range(0,n):
    for j in range(0,m):
        L[i,j]+=1 # on additionne 1 au coeff i,j de L
        print("coef (", i+1,",",j+1,") de L = ", L[i,j])
    # fin for j
# fin for i
        
```

Pour récupérer la ligne $i$ (ou la colonne $j$) de una matrice A, on peut utiliser la syntaxe A[i,:] (ou A[:,j]) (enlever les commentaires des lignes d'affichage et exécuter) : 

```python
print(L[0,:])  #1ère ligne de L
```

```python
print(L[:,1])  #2ème colonne de L (mais donné comme une ligne)
```

On peut aussi récupérer une partie de la ligne $i$ ou de la colonne $j$ (enlever les commentaires des lignes d'affichage et exécuter) : 

```python
print(L[0:2,1])  #on récupère les 2 premiers coefficients de la 2ème colonne de L 
```

b)  Avec np.ones et np.diag écrire une fonction pour créer la matrice tridiagonale carrée d'ordre $n$ suivante

$
   T= \left[\begin{array}{rrrrrr}
           2 & -1 & 0 & \cdots & \cdots & 0 \\
           -1 & 2 & -1 &  0 & \cdots   & 0\\
           0 & -1 & 2 & -1 & 0 & \cdots  \\
             & & \ddots &\ddots &\ddots &  \\
             0 &\cdots &\cdots&  -1 & 2 & -1  \\
           0 &\cdots &\cdots& 0& -1 & 2 &  
         \end{array}\right]%.
$. 

La fonction aura comme argument $n$ (la taille de la matrice)


```python
def tridiagonale(n):
    d = np.ones(n)
    T = 2.*np.diag(d) - np.diag(d[0:n-1],1) - np.diag(d[0:n-1],-1)  
    return T

T = tridiagonale(10)
print(T)
```

## Exercice 3

 La matrice de Hilbert $H$ de taille $n$ est définie par 


$\quad  (H)_{ij} = \frac{1}{i+j-1}, \quad 1\leq i,j \leq n.$

Par exemple, pour $n=4$, 
$
   H= \left[\begin{array}{rrrr}
           1 & 1/{2} & 1/{3} & 1/{4} \\
           1/{2} & 1/{3} & 1/{4} & 1/{5} \\
           1/{3} & 1/{4} & 1/{5} & 1/{6} \\
           1/{4} & 1/{5} & 1/{6} & 1/{7} \\
         \end{array}\right]%.
$. 
Ecrire une fonction python pour construire la matrice de Hilbert de taille $n$. 

On pourra compléter la fonction ci-dessous (noter que les indices commencent à 0)

```python
def mathilbert(n):
    H = np.zeros((n,n)) # initialisation avec un matrice nule de taille nxn
    for i in range(0,n): # compléter la paranthèse  
        for j in range(0,n):#... à compléter
            H[i,j] = 1/(i+j+1) # car i,j >= 0, on decale de 1, et (i+1)+(i+j)-1 = i+j+1
    return H

mathilbert(4)
```

b) Est-ce que l'on pourrait écrire une fonction similaire mais avec moins de boucles for ? (essayer ci-dessous)
