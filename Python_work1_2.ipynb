## Erreur de représentation, arithmétique en virgule flottante, opérations risquées

Soit $F\subset \mathbb{R}$ un ensemble de nombres  à virgule flottante (avec une mantise de taille $t$). Soit 
$fl : \mathbb{R} \to F$ l'application qui à $x\in \mathbb{R}$ associe le nombre à virgule floattante obtenu en utilisant un arrondi. On a alors

$\quad \displaystyle\frac{|fl(x) -x|}{|x|} \le \frac{\epsilon_M}{2}.$

Les opérations arithmétiques $op$ ($op = +,-,\times,\div$) dans l'ensemble $F$ sont définies de la manière suivante : Soient $a,b\in F$

$\quad a\, op\, b = fl(a\,op_\mathbb{R} \,b)$.

où $op_\mathbb{R}$ dénote l'opération arithmétique exacte (c.-à-d. dans $\mathbb{R}$). 

Si $x,y\in\mathbb{R}$ alors, l'opération dans $F$ sera

$\quad x+y = fl(fl(x) + fl(y))$

On voit que, dû a l'erreur de représentation, chaque opération arithmétique introduit alors une petite erreur. 
Il est important  noter que si une séquence d'opérations est réalisée alors l'erreur totale n'est pas nécessairement petite.

Il y a aussi des opérations risquées, par exemple :

- Additionner deux nombres dont les nombres de grandeurs sont très différents

- Soustraire deux nombres presqu'identiques (élimination par soustraction des chiffres significatifs)

- Quand $a\,op\,b\notin F$ (les exceptions).

## Exercice 9. (Algorithme d'Horner)

L'évaluation directe d'une fonction polynomiale nécessite le calcul des puissances de $x$. En remarquant que le polynôme de l'exercice 8 peut s'écrire de la manière suivante

$\quad f(x) = -1 + x(7 + x(-21 + x(35 + x(-35 + x(21 + x(-7 + x)))))),$

on voit que l'on pourrait évaluer la fonction $f$, sans calculer les puissances de $x$, en effectuant des opération intermediaires (en partant de la dernière parenthèse à l'intérieur). 

L'algorithme suivant présente cette méthode de manière générale :

1) Soient $a_i$, $i=0,\dots,n$, les coefficients d'un polynôme P de degré $n$; Soit $x$ un point d'abscisse.

2) Initialisation : $p = a_n$ ( $p$ contiendra la valeur $P(x)$)

3) Pour $k=n,n-1,\dots,2,1$ effectuer

$\quad \quad p = a_{k-1} + x \times p$

4) Ecrire $P(x) = p$.

Cette méthode à l'avantage d'évaluer la fonction polynomiale d'une manière plus rapide (car il y a moins d'opérations à effectuer) et plus stable (du point de vue de l'arithmétique flottante).
